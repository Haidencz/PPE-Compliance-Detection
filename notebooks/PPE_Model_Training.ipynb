```python
from pathlib import Path
from zipfile import ZipFile


archive_path = Path("/content/ppe_dataset.zip")
dataset_path = Path("/content/ppe_dataset")

with ZipFile(archive_path, "r") as archive:
    archive.extractall(dataset_path)

print("Dataset extracted to:", dataset_path)
```

```python
for split in ["train", "valid", "test"]:
    image_count = len(list((dataset_path / split / "images").glob("*")))
    label_count = len(list((dataset_path / split / "labels").glob("*.txt")))

    print(f"{split}: {image_count} images, {label_count} labels")

print("\nDataset configuration:")
print((dataset_path / "data.yaml").read_text())
```

```python
print("Top-level contents:")

for item in dataset_path.iterdir():
    print(repr(item.name), "directory" if item.is_dir() else "file")

print("\nFirst 30 extracted paths:")

for index, item in enumerate(dataset_path.rglob("*")):
    print(item.relative_to(dataset_path))

    if index == 29:
        break
```

```python
import shutil
from pathlib import Path
from zipfile import ZipFile


dataset_path = Path("/content/ppe_dataset_fixed")
dataset_path.mkdir(parents=True, exist_ok=True)

with ZipFile(archive_path, "r") as archive:
    for member in archive.infolist():
        normalised_name = member.filename.replace("\\", "/")
        relative_path = Path(normalised_name)

        if ".." in relative_path.parts:
            raise ValueError(f"Unsafe archive path: {normalised_name}")

        destination = dataset_path / relative_path

        if normalised_name.endswith("/"):
            destination.mkdir(parents=True, exist_ok=True)
            continue

        destination.parent.mkdir(parents=True, exist_ok=True)

        with archive.open(member) as source:
            with destination.open("wb") as target:
                shutil.copyfileobj(source, target)

print("Corrected dataset extracted to:", dataset_path)
```

```python
for split in ["train", "valid", "test"]:
    image_count = len(list((dataset_path / split / "images").glob("*")))
    label_count = len(list((dataset_path / split / "labels").glob("*.txt")))

    print(f"{split}: {image_count} images, {label_count} labels")
```

```python
from ultralytics import YOLO


model = YOLO("yolo11n.pt")

smoke_results = model.train(
    data=str(dataset_path / "data.yaml"),
    epochs=1,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="/content/runs",
    name="ppe_smoke_test",
    seed=42
)
```

```python
baseline_model = YOLO("yolo11n.pt")

baseline_results = baseline_model.train(
    data=str(dataset_path / "data.yaml"),
    epochs=50,
    patience=10,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="/content/runs",
    name="ppe_baseline",
    seed=42,
    plots=True
)
```
